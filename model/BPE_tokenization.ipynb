```python
from tokenizers import ByteLevelBPETokenizer
import csv
import sys
import os
```

```python
csv.field_size_limit(sys.maxsize)

with open("dataset/wikipedia_1M_1.csv", encoding="utf-8", newline="") as source, \
     open("dataset/wikipedia_text.txt", "w", encoding="utf-8") as target:
    reader = csv.DictReader(source)
    for row in reader:
        text = row["text"].strip()
        if text:
            target.write(text + "\n")
```

```python
tokenizer = ByteLevelBPETokenizer()
```

```python
tokenizer.train(
    files=["dataset/wikipedia_text.txt"],
    vocab_size=30_000,
    min_frequency=5,
    special_tokens=["<s>", "<pad>", "</s>", "<unk>", "<mask>"],
)
```

```python
test_output = tokenizer.encode("Hello, world!")
print(test_output.tokens)
```

```python
OUTPUT_DIR = "tokenizer"
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
tokenizer.save_model(OUTPUT_DIR)
```
